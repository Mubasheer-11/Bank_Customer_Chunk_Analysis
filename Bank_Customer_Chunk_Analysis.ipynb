```python
# Loading the dataset using pandas

import pandas as pd

df = pd.read_csv('Bank Customer Churn Prediction.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
# Summary statistics using .describe()
df.describe(include='all')
```

```python
# Checking if missing data or null values are present in the dataset

df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(r"[-_]+", "", regex=True)
```

```python
df.columns
```

```python
print(df.columns.tolist())
```

```python
# Imputing missing values in Review Rating column with the median rating of the product category

df['creditscore'] = df['creditscore'].fillna(
    df['creditscore'].median()
)
```

```python
df.isnull().sum()
```

```python
# create a new column age_group
labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels = labels)
```

```python
df[['age','age_group']].head(10)
```

```python
# Create new column: tenure_category

df['tenure_category'] = pd.cut(
    df['tenure'],
    bins=[-1, 2, 5, 8, 10],
    labels=['New', 'Medium', 'Long-term', 'Loyal']
)
```

```python
df[['tenure', 'tenure_category']].head(10)
```

```python
df[['creditcard', 'activemember']].head(10)
```

```python
df['customerid'].is_unique
```

```python
# Dropping customer ID column

df = df.drop('customerid', axis=1)
```
